```python
import pandas as pd

# データの読み込み
df = pd.read_parquet(
    "/Users/reiyamamoto/Documents/GitHub/DataScienceCompetition/parquet_data/1571_ja.parquet",
    engine="fastparquet"
)

# 1. データの全体像（列の数や名前）を出力する
print("▼データの情報")
df.info()

# 2. データの中身（最初の5行）を出力する
print("\n▼データの中身（最初の5行）")
print(df.head())
```

```python
#とりあえずクロス集計してみる
# 1. 分析に使いたい列をリストアップ
cols = [
    'w1p_昨年の世帯収入', 
    'w1c_生活時間：（学習塾）週に何回行くか'
]

# 2. 必要な列だけを切り出した新しい表（データフレーム）を作る
df_theme1 = df[cols].copy()

# 3. アンケート特有のダミー数値（7777: 非該当、9999: 無回答など）を除外する
# ※ 有効な選択肢の番号だけを残すための簡単な処理です
df_theme1 = df_theme1[
    (df_theme1['w1p_昨年の世帯収入'] < 100) & 
    (df_theme1['w1c_生活時間：（学習塾）週に何回行くか'] < 100)
]

# 4. クロス集計表を作成（行：収入、列：塾の回数）
# normalize='index' を入れることで、各収入層ごとの「割合（％）」を出せます
cross_table = pd.crosstab(
    df_theme1['w1p_昨年の世帯収入'], 
    df_theme1['w1c_生活時間：（学習塾）週に何回行くか'], 
    normalize='index'
) * 100

# 結果を小数点第1位で丸めて表示
round(cross_table, 1)
```

```python
#＜悩み「習い事や塾・教材選び」をもつ保護者が、子供を習い事に通わせている割合は高いのか？＞

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib

# 分析に必要な変数の抽出
columns_nayami = [
    'w1p_悩みや気がかり：習い事や塾・教材選び', 
    'w1p_習い事やスポーツクラブに通っているか'
]

df_nayami = df[columns_nayami].copy()

# 「習い事に通っているか」の無効な回答（7777など）をNaNにして除外
df_nayami['w1p_習い事やスポーツクラブに通っているか'] = df_nayami['w1p_習い事やスポーツクラブに通っているか'].replace([7777, 8888, 9999], np.nan)
df_nayami = df_nayami.dropna(subset=['w1p_習い事やスポーツクラブに通っているか'])

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
# ① 習い事に通っているかどうかのフラグ作成 (1:行っている->1, 2:行っていない->0)
df_nayami['has_lesson'] = df_nayami['w1p_習い事やスポーツクラブに通っているか'].apply(lambda x: 1 if x == 1 else 0)

# ② 「習い事や塾・教材選び」の悩みフラグ作成
# 複数回答の項目は、選んでいれば「1」や「項目番号(28など)」、選んでいなければ「0」や「NaN」になります。
# 値が1以上で、かつ異常値(7777等)ではない場合に「1(悩みあり)」とする安全な処理を行います。
def check_nayami(x):
    if pd.isna(x) or x in [7777, 8888, 9999] or x == 0:
        return 0 # 悩みなし
    else:
        return 1 # 悩みあり

df_nayami['has_nayami'] = df_nayami['w1p_悩みや気がかり：習い事や塾・教材選び'].apply(check_nayami)
df_nayami['nayami_label'] = df_nayami['has_nayami'].map({0: '悩みなし', 1: '悩みあり'})

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
# 「悩みあり／なし」ごとの「習い事実施割合」を算出
nayami_lesson_mean = df_nayami.groupby('nayami_label')['has_lesson'].mean().reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(x='nayami_label', y='has_lesson', data=nayami_lesson_mean, palette='Set2')
plt.title('「習い事や塾・教材選び」の悩み有無と、実際の習い事実施割合')
plt.xlabel('保護者の悩み')
plt.ylabel('子どもが習い事に通っている割合')
plt.ylim(0, 1)
plt.show()

# 数値データの表示
print("■ 「習い事や塾・教材選び」の悩み有無ごとの、習い事実施割合")
nayami_lesson_mean.columns = ['保護者の悩み', '習い事実施割合']
display(nayami_lesson_mean)
```

```python
#＜ひとり親家庭（シングルマザー、シングルファザー問題）では、機会損失が起こりやすいのでは？＞

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib  # 文字化け防止ツールを読み込み

# =========================================================
# 仮説検証：ひとり親世帯における「習い事・塾」の機会格差
# =========================================================

# 分析に必要な変数の抽出
columns_single = [
    'w1p_配偶者の有無', # 1: いる(ふたり親), 2: いない(ひとり親)
    'w1p_習い事やスポーツクラブに通っているか', # 1: 行っている, 2: 行っていない
    'w1p_現在学習塾に行っているか' # 1: 行っている, 2: 行っていない
]

df_single = df[columns_single].copy()

# 欠損値の処理 (7777, 8888, 9999 をまとめて NaN に置換して除外)
for col in columns_single:
    df_single[col] = df_single[col].replace([7777, 8888, 9999], np.nan)

df_single = df_single.dropna()
print("有効データ数:", len(df_single))

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
# ① 世帯形態のラベル作成
df_single['family_type'] = df_single['w1p_配偶者の有無'].map({1.0: '配偶者あり（ふたり親）', 2.0: '配偶者なし（ひとり親）'})

# ② 実施フラグの作成 (1:行っている->1, 2:行っていない->0)
df_single['has_lesson'] = df_single['w1p_習い事やスポーツクラブに通っているか'].apply(lambda x: 1 if x == 1 else 0)
df_single['has_juku'] = df_single['w1p_現在学習塾に行っているか'].apply(lambda x: 1 if x == 1 else 0)

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
from IPython.display import display

# 世帯形態ごとの実施割合を算出
summary_single = df_single.groupby('family_type')[['has_lesson', 'has_juku']].mean().reset_index()

# グラフで描きやすいようにデータ形式を変換（Long形式）
df_melted_single = pd.melt(summary_single, id_vars=['family_type'], 
                           value_vars=['has_lesson', 'has_juku'],
                           var_name='activity_type', value_name='participation_rate')

# ラベルを日本語に変換
df_melted_single['activity_type'] = df_melted_single['activity_type'].map({
    'has_lesson': '習い事・スポーツクラブ', 
    'has_juku': '学習塾'
})

# 棒グラフの描画
plt.figure(figsize=(9, 6))
sns.barplot(x='activity_type', y='participation_rate', hue='family_type', data=df_melted_single, palette='Set2')
plt.title('世帯形態（ひとり親・ふたり親）による「習い事・塾」の機会格差')
plt.xlabel('活動の種類')
plt.ylabel('通っている子どもの割合')
plt.ylim(0, 1)

# グラフ上にパーセンテージを表示する工夫
ax = plt.gca()
for p in ax.patches:
    ax.annotate(f"{p.get_height():.1%}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=10, color='black', xytext=(0, 5), textcoords='offset points')

plt.legend(title='世帯形態')
plt.show()

# 数値データの表示
print("■ 分析結果：世帯形態ごとの実施割合（%）")
summary_single.columns = ['世帯形態', '習い事実施割合', '学習塾実施割合']
# 見やすくするためにパーセント表示に変換
summary_single['習い事実施割合'] = (summary_single['習い事実施割合'] * 100).round(1).astype(str) + '%'
summary_single['学習塾実施割合'] = (summary_single['学習塾実施割合'] * 100).round(1).astype(str) + '%'
display(summary_single)
```

```python
#＜ひとり親家庭では、子供の最終学歴が中学校や高校で止まってしまうのでは？＞

# =========================================================
# 仮説検証：ひとり親家庭における「進学希望（高卒以下）」の格差
# =========================================================

# 分析に必要な変数の抽出
columns_edu = [
    'w1p_配偶者の有無',        # 1: いる(ふたり親), 2: いない(ひとり親)
    'w1c_将来の希望進学段階'   # 子ども自身の進学希望
]

df_edu = df[columns_edu].copy()

# 欠損値の処理 (7777, 8888, 9999 をまとめて NaN に置換)
for col in columns_edu:
    df_edu[col] = df_edu[col].replace([7777, 8888, 9999], np.nan)

# 選択肢 7(その他) と 8(まだ決めていない) を除外して、具体的な進学目標がある層で比較
df_edu = df_edu[df_edu['w1c_将来の希望進学段階'].isin([1, 2, 3, 4, 5, 6])]
df_edu = df_edu.dropna()
print("有効データ数:", len(df_edu))

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
# ① 世帯形態のラベル作成
df_edu['family_type'] = df_edu['w1p_配偶者の有無'].map({1.0: 'ふたり親世帯', 2.0: 'ひとり親世帯'})

# ② 「中学校・高校まで（高卒以下）」を希望しているかどうかのフラグを作成
# 1: 中学校まで, 2: 高校まで -> 1 (高卒以下を希望)
# 3: 専門・各種学校, 4: 短大, 5: 大学, 6: 大学院 -> 0 (それ以上の進学を希望)
df_edu['aim_high_school_or_less'] = df_edu['w1c_将来の希望進学段階'].apply(lambda x: 1 if x in [1, 2] else 0)

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
# 世帯形態ごとの「高卒以下希望」の割合を算出
edu_summary = df_edu.groupby('family_type')['aim_high_school_or_less'].mean().reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(x='family_type', y='aim_high_school_or_less', data=edu_summary, palette='Set1')
plt.title('世帯形態による「中学校・高校までの進学」を希望する子どもの割合')
plt.xlabel('世帯形態')
plt.ylabel('「中学・高校まで」を希望する割合')
plt.ylim(0, 0.5)  # 最大50%に設定（結果に応じて調整可能）

# グラフ上にパーセンテージを表示
ax = plt.gca()
for p in ax.patches:
    ax.annotate(f"{p.get_height():.1%}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=12, color='black', xytext=(0, 5), textcoords='offset points')

plt.show()

# 数値データの表示
print("■ 分析結果：世帯形態ごとの「中学校・高校まで」の進学希望割合（%）")
edu_summary.columns = ['世帯形態', '中学・高校までの進学希望割合']
edu_summary['中学・高校までの進学希望割合'] = (edu_summary['中学・高校までの進学希望割合'] * 100).round(1).astype(str) + '%'
display(edu_summary)
```

```python
#＜親の人数と子供が希望する最終学歴＞

# =========================================================
# 修正版：「中高生」に絞った世帯形態と進学希望のクロス分析
# =========================================================

# 今度は「学年」の列も一緒に抽出します
columns_edu_corr = [
    'w6p_ 配偶者の有無',       # 1:いる, 2:いない
    'w6c_ 将来の希望進学段階',  
    'w6学年'                  # 1〜6:小学, 7〜9:中学, 10〜12:高校
]

df_edu_corr = df[columns_edu_corr].copy()

# 1. 「中学生・高校生（学年コード7以上）」だけにデータを絞り込む
df_edu_corr = df_edu_corr[df_edu_corr['w6学年'] >= 7]

# 2. 欠損値と未定の除外
for col in columns_edu_corr:
    df_edu_corr[col] = df_edu_corr[col].replace([7777, 8888, 9999], np.nan)
df_edu_corr['w6c_ 将来の希望進学段階'] = df_edu_corr['w6c_ 将来の希望進学段階'].replace([8, 9], np.nan)
df_edu_corr = df_edu_corr.dropna()

print(f"中高生に絞った有効データ数: {len(df_edu_corr)}人")

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
df_edu_corr['family_type'] = df_edu_corr['w6p_ 配偶者の有無'].map({1.0: 'ふたり親世帯', 2.0: 'ひとり親世帯'})

def group_education(x):
    if x in [1, 2]:
        return '1. 中学・高校まで'
    elif x in [3, 4]:
        return '2. 専門学校・短大まで'
    elif x in [5, 6, 7]:
        return '3. 大学・大学院まで'
    return np.nan

df_edu_corr['edu_aspiration'] = df_edu_corr['w6c_ 将来の希望進学段階'].apply(group_education)

# ---------------------------------------------------------
# グラフの描画
# ---------------------------------------------------------
cross_table_corr = pd.crosstab(df_edu_corr['family_type'], df_edu_corr['edu_aspiration'], normalize='index') * 100

ax = cross_table_corr.plot(kind='barh', stacked=True, figsize=(10, 4), color=['#ff9999', '#ffcc99', '#99ccff'])
plt.title('【中高生限定】世帯形態による「子どもの進学希望」の格差')
plt.xlabel('割合 (%)')
plt.ylabel('世帯形態')
plt.legend(title='子どもの希望進学段階', bbox_to_anchor=(1.05, 1), loc='upper left')

for n, x in enumerate([*cross_table_corr.index.values]):
    for (proportion, x_loc) in zip(cross_table_corr.loc[x], cross_table_corr.loc[x].cumsum()):
        if proportion > 5:
            plt.text(x=x_loc - (proportion / 2), y=n, s=f'{proportion:.1f}%', 
                     color="black", fontsize=10, ha="center", va="center")
plt.tight_layout()
plt.show()

display(round(cross_table_corr, 1))
```

```python
#＜親が忙しくて「子どもと過ごす時間」が取れないことが、子どもの「学習習慣の獲得」という機会を奪っていないか？＞

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib

# =========================================================
# 分析：親の「タイムプア」が子どもの学習習慣に与える影響
# =========================================================

# 1. 分析に必要な変数の抽出（Wave 6のデータを使用）
columns_timepoor = [
    'w6p_ あなた：ふだん子どもと過ごす時間',              # タイムプアの指標
    'w6p_ 勉強かかわり：勉強の計画の仕方を教える',        # 親のサポート
    'w6c_ 勉強方法：計画をたてて勉強する',                # 子どもの学習習慣
    'w6c_ 性格・特性：上手な勉強のしかたがわからない'     # 学習の迷子度合い
]

df_timepoor = df[columns_timepoor].copy()

# 欠損値の処理 (7777, 8888, 9999 をまとめて NaN に置換して除外)
for col in columns_timepoor:
    df_timepoor[col] = df_timepoor[col].replace([7777, 8888, 9999], np.nan)

df_timepoor = df_timepoor.dropna()
print(f"有効データ数: {len(df_timepoor)}人")

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
# スコアの反転（数値が大きいほど「よくする」「あてはまる」になるよう変換: 5 - 元の値）
# ※「勉強のしかたがわからない」はそのまま（数値が大きいほど「わからない」状態）
df_timepoor['teach_planning'] = 5 - df_timepoor['w6p_ 勉強かかわり：勉強の計画の仕方を教える']
df_timepoor['child_planning'] = 5 - df_timepoor['w6c_ 勉強方法：計画をたてて勉強する']
df_timepoor['dont_know_how'] = 5 - df_timepoor['w6c_ 性格・特性：上手な勉強のしかたがわからない']

# 親が子どもと過ごす時間を分かりやすくカテゴリ化（仮の分割例）
# ※実際の値の分布に合わせて調整してください（例：1〜2を短い、3を普通、4以上を長いなど）
def group_time(x):
    if x <= 2:
        return '1. 短い'
    elif x == 3:
        return '2. 普通'
    else:
        return '3. 長い'

df_timepoor['time_spent_group'] = df_timepoor['w6p_ あなた：ふだん子どもと過ごす時間'].apply(group_time)

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
from IPython.display import display

# グループごとの平均値を算出
summary_timepoor = df_timepoor.groupby('time_spent_group')[['teach_planning', 'child_planning', 'dont_know_how']].mean().reset_index()

# グラフで描きやすいようにデータ形式を変換（Long形式）
df_melted_timepoor = pd.melt(summary_timepoor, id_vars=['time_spent_group'], 
                             value_vars=['teach_planning', 'child_planning', 'dont_know_how'],
                             var_name='metric_type', value_name='mean_score')

# ラベルを日本語に変換
label_map_time = {
    'teach_planning': '親：計画の立て方を教える',
    'child_planning': '子：計画を立てて勉強する',
    'dont_know_how': '子：勉強のしかたがわからない'
}
df_melted_timepoor['metric_type'] = df_melted_timepoor['metric_type'].map(label_map_time)

# 折れ線グラフの描画（傾向を見るためにLineplotを使用）
plt.figure(figsize=(10, 6))
sns.lineplot(x='time_spent_group', y='mean_score', hue='metric_type', data=df_melted_timepoor, marker='o', linewidth=2.5, markersize=10)

plt.title('親の「タイムプア（過ごす時間）」が子どもの学習習慣に与える影響')
plt.xlabel('親が子どもと過ごす時間の長さ')
plt.ylabel('平均スコア (高いほどあてはまる)')
plt.ylim(1.5, 3.5) # Y軸の範囲を見やすく調整
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='評価項目', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()

# 数値データの表示
print("■ 分析結果：親と過ごす時間の長さ別、各スコアの平均")
summary_timepoor.columns = ['親と過ごす時間の長さ', '親が計画の立て方を教える頻度', '子が計画を立てて勉強する頻度', '子が勉強のしかたがわからない度合い']
display(summary_timepoor)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib

# =========================================================
# 分析：親の「時間的貧困」の実態と子どもの意欲への影響
# =========================================================

# 1. 分析に必要な変数の抽出
# ※列名に「平日」や「休日」の区別がある場合は、実際の列名に合わせて適宜修正してください
columns_time = [
    'w6p_ 配偶者の有無', 
    'w6p_ あなた：ふだん子どもと過ごす時間', 
    'w6p_ 配偶者：ふだん子どもと過ごす時間',
    'w6c_ 性格・特性：難しいことや新しいことにいつも挑戦したい' 
]
df_time = df[columns_time].copy()

# ---------------------------------------------------------
# ① 仮説の検証：「ひとり親は配偶者の回答を飛ばしているか？」
# ---------------------------------------------------------
single_spouse_time = df_time[df_time['w6p_ 配偶者の有無'] == 2.0]['w6p_ 配偶者：ふだん子どもと過ごす時間']

print("■ ひとり親世帯における「配偶者の過ごす時間」の入力値")
print("（※想定通りなら、ほとんどが『8888(非該当)』になっているはずです）")
print(single_spouse_time.value_counts(dropna=False))
print("-" * 50)

# ---------------------------------------------------------
# データの加工と欠損値処理
# ---------------------------------------------------------
# ② 欠損値の処理 (7777, 8888, 9999 をまとめて NaN に置換)
for col in columns_time:
    df_time[col] = df_time[col].replace([7777, 8888, 9999], np.nan)

# ★超重要ポイント★
# ここで df_time.dropna() をしてしまうと、配偶者の時間がNaNになっている
# 「ひとり親世帯」が全て消えてしまうため、条件を指定して除外します。
df_time = df_time.dropna(subset=[
    'w6p_ 配偶者の有無', 
    'w6p_ あなた：ふだん子どもと過ごす時間',
    'w6c_ 性格・特性：難しいことや新しいことにいつも挑戦したい'
])

# ③ 世帯形態のラベル作成
df_time['family_type'] = df_time['w6p_ 配偶者の有無'].map({1.0: 'ふたり親世帯', 2.0: 'ひとり親世帯'})

# ④ 子どもの挑戦意欲スコアの調整（反転）
df_time['challenge_spirit'] = 5 - df_time['w6c_ 性格・特性：難しいことや新しいことにいつも挑戦したい']

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
from IPython.display import display

# グラフ①：世帯形態による「あなたが子どもと過ごす時間」の違い
plt.figure(figsize=(8, 5))
# カテゴリ値（大きいほど時間が長い）の平均を比較します
sns.barplot(x='family_type', y='w6p_ あなた：ふだん子どもと過ごす時間', data=df_time, palette='Set2')
plt.title('世帯形態による「親（回答者）が子どもと過ごす時間」の比較')
plt.xlabel('世帯形態')
plt.ylabel('過ごす時間スコア (高いほど長い)')
plt.show()

# グラフ②：親と過ごす時間と「子どもの挑戦意欲」の関連
plt.figure(figsize=(9, 5))
sns.barplot(x='w6p_ あなた：ふだん子どもと過ごす時間', y='challenge_spirit', data=df_time, palette='Blues')
plt.title('親が子どもと過ごす時間と「子どもの挑戦意欲」の関連')
plt.xlabel('親が子どもと過ごす時間（スコア：大きいほど長い）')
plt.ylabel('子どもの挑戦意欲スコア平均')
plt.ylim(2.0, 3.5) # 差を見やすく調整
plt.show()

# 数値データの表示（グラフ②）
print("■ 分析結果：親と過ごす時間ごとの、子どもの挑戦意欲スコア平均")
time_challenge_mean = df_time.groupby('w6p_ あなた：ふだん子どもと過ごす時間')['challenge_spirit'].mean().reset_index()
display(time_challenge_mean)
```

```python
import pandas as pd
import numpy as np
from IPython.display import display

# =========================================================
# データ確認：Wave1〜Wave7における「子どもの学年分布」の推移
# =========================================================

# Wave1〜7の学年列名をリスト化
grade_columns = [f'w{i}学年' for i in range(1, 8)]

# 結果をまとめるための空のデータフレームを作成
grade_distribution = pd.DataFrame()

for col in grade_columns:
    # 対象の列がデータフレームに存在するか確認
    if col in df.columns:
        # コピーして無効な値(7777等)をNaNに置換
        temp_series = df[col].replace([7777, 8888, 9999], np.nan)
        
        # 割合を集計してパーセント（%）に変換
        counts = temp_series.value_counts(normalize=True).sort_index() * 100
        grade_distribution[col] = counts

# インデックス（学年コード）を見やすいように整数型へ変換
grade_distribution.index = grade_distribution.index.astype(int)

# データが存在しない箇所（NaN）を0で埋め、小数点第1位に丸める
grade_distribution = grade_distribution.fillna(0).round(1)

# 表を見やすくするために「%」を追加し、0.0%の箇所は「-」で見やすくする
formatted_distribution = grade_distribution.astype(str) + "%"
formatted_distribution = formatted_distribution.replace("0.0%", "-")

# 行と列のラベルを分かりやすく変更
formatted_distribution.index.name = '学年コード'
formatted_distribution.columns = [f'Wave {i}' for i in range(1, 8)]

print("■ Wave1〜Wave7における子どもの学年分布（構成比の推移）")
display(formatted_distribution)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib

# =========================================================
# グラフ化：Wave1〜Wave7における「子どもの学年分布」のヒートマップ（件数版）
# =========================================================

# 1. データの再集計（件数として集計）
grade_columns = [f'w{i}学年' for i in range(1, 8)]
grade_distribution_count = pd.DataFrame()

for col in grade_columns:
    if col in df.columns:
        temp_series = df[col].replace([7777, 8888, 9999], np.nan)
        # normalize=Trueを外し、純粋な件数をカウントします
        counts = temp_series.value_counts().sort_index()
        grade_distribution_count[col] = counts

grade_distribution_count.index = grade_distribution_count.index.astype(int)
grade_distribution_count = grade_distribution_count.fillna(0)
grade_distribution_count.columns = [f'Wave {i}' for i in range(1, 8)]
grade_distribution_count.index.name = '学年コード'

# 2. ヒートマップの描画
plt.figure(figsize=(10, 8))

# annot=Trueで数値を表示、fmt=".0f"で小数点なしの整数に指定
sns.heatmap(grade_distribution_count, cmap='Blues', annot=True, fmt=".0f", 
            cbar_kws={'label': '件数（人）'}, linewidths=.5)

plt.title('Wave1〜Wave7における子どもの学年分布の推移（実数）')
plt.xlabel('調査の回数 (Wave)')
plt.ylabel('学年コード (1:小1 〜 12:高3)')

# Y軸の数字を見やすく水平にする
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```
