```python
import pandas as pd

# データの読み込み
df = pd.read_parquet(
    "/Users/reiyamamoto/Documents/GitHub/DataScienceCompetition/parquet_data/1571_ja.parquet",
    engine="fastparquet"
)

# 1. データの全体像（列の数や名前）を出力する
print("▼データの情報")
df.info()

# 2. データの中身（最初の5行）を出力する
print("\n▼データの中身（最初の5行）")
print(df.head())
```

```python
#とりあえずクロス集計してみる
# 1. 分析に使いたい列をリストアップ
cols = [
    'w1p_昨年の世帯収入', 
    'w1c_生活時間：（学習塾）週に何回行くか'
]

# 2. 必要な列だけを切り出した新しい表（データフレーム）を作る
df_theme1 = df[cols].copy()

# 3. アンケート特有のダミー数値（7777: 非該当、9999: 無回答など）を除外する
# ※ 有効な選択肢の番号だけを残すための簡単な処理です
df_theme1 = df_theme1[
    (df_theme1['w1p_昨年の世帯収入'] < 100) & 
    (df_theme1['w1c_生活時間：（学習塾）週に何回行くか'] < 100)
]

# 4. クロス集計表を作成（行：収入、列：塾の回数）
# normalize='index' を入れることで、各収入層ごとの「割合（％）」を出せます
cross_table = pd.crosstab(
    df_theme1['w1p_昨年の世帯収入'], 
    df_theme1['w1c_生活時間：（学習塾）週に何回行くか'], 
    normalize='index'
) * 100

# 結果を小数点第1位で丸めて表示
round(cross_table, 1)
```
