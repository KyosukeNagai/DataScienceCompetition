```python
import pandas as pd

# データの読み込み
df = pd.read_parquet(
    "/Users/reiyamamoto/Documents/GitHub/DataScienceCompetition/parquet_data/1571_ja.parquet",
    engine="fastparquet"
)

# 1. データの全体像（列の数や名前）を出力する
print("▼データの情報")
df.info()

# 2. データの中身（最初の5行）を出力する
print("\n▼データの中身（最初の5行）")
print(df.head())
```

```python
#とりあえずクロス集計してみる
# 1. 分析に使いたい列をリストアップ
cols = [
    'w1p_昨年の世帯収入', 
    'w1c_生活時間：（学習塾）週に何回行くか'
]

# 2. 必要な列だけを切り出した新しい表（データフレーム）を作る
df_theme1 = df[cols].copy()

# 3. アンケート特有のダミー数値（7777: 非該当、9999: 無回答など）を除外する
# ※ 有効な選択肢の番号だけを残すための簡単な処理です
df_theme1 = df_theme1[
    (df_theme1['w1p_昨年の世帯収入'] < 100) & 
    (df_theme1['w1c_生活時間：（学習塾）週に何回行くか'] < 100)
]

# 4. クロス集計表を作成（行：収入、列：塾の回数）
# normalize='index' を入れることで、各収入層ごとの「割合（％）」を出せます
cross_table = pd.crosstab(
    df_theme1['w1p_昨年の世帯収入'], 
    df_theme1['w1c_生活時間：（学習塾）週に何回行くか'], 
    normalize='index'
) * 100

# 結果を小数点第1位で丸めて表示
round(cross_table, 1)
```

```python
#＜悩み「習い事や塾・教材選び」をもつ保護者が、子供を習い事に通わせている割合は高いのか？＞

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib

# 分析に必要な変数の抽出
columns_nayami = [
    'w1p_悩みや気がかり：習い事や塾・教材選び', 
    'w1p_習い事やスポーツクラブに通っているか'
]

df_nayami = df[columns_nayami].copy()

# 「習い事に通っているか」の無効な回答（7777など）をNaNにして除外
df_nayami['w1p_習い事やスポーツクラブに通っているか'] = df_nayami['w1p_習い事やスポーツクラブに通っているか'].replace([7777, 8888, 9999], np.nan)
df_nayami = df_nayami.dropna(subset=['w1p_習い事やスポーツクラブに通っているか'])

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
# ① 習い事に通っているかどうかのフラグ作成 (1:行っている->1, 2:行っていない->0)
df_nayami['has_lesson'] = df_nayami['w1p_習い事やスポーツクラブに通っているか'].apply(lambda x: 1 if x == 1 else 0)

# ② 「習い事や塾・教材選び」の悩みフラグ作成
# 複数回答の項目は、選んでいれば「1」や「項目番号(28など)」、選んでいなければ「0」や「NaN」になります。
# 値が1以上で、かつ異常値(7777等)ではない場合に「1(悩みあり)」とする安全な処理を行います。
def check_nayami(x):
    if pd.isna(x) or x in [7777, 8888, 9999] or x == 0:
        return 0 # 悩みなし
    else:
        return 1 # 悩みあり

df_nayami['has_nayami'] = df_nayami['w1p_悩みや気がかり：習い事や塾・教材選び'].apply(check_nayami)
df_nayami['nayami_label'] = df_nayami['has_nayami'].map({0: '悩みなし', 1: '悩みあり'})

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
# 「悩みあり／なし」ごとの「習い事実施割合」を算出
nayami_lesson_mean = df_nayami.groupby('nayami_label')['has_lesson'].mean().reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(x='nayami_label', y='has_lesson', data=nayami_lesson_mean, palette='Set2')
plt.title('「習い事や塾・教材選び」の悩み有無と、実際の習い事実施割合')
plt.xlabel('保護者の悩み')
plt.ylabel('子どもが習い事に通っている割合')
plt.ylim(0, 1)
plt.show()

# 数値データの表示
print("■ 「習い事や塾・教材選び」の悩み有無ごとの、習い事実施割合")
nayami_lesson_mean.columns = ['保護者の悩み', '習い事実施割合']
display(nayami_lesson_mean)
```

```python
#＜ひとり親家庭（シングルマザー、シングルファザー問題）では、機会損失が起こりやすいのでは？＞

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib  # 文字化け防止ツールを読み込み

# =========================================================
# 仮説検証：ひとり親世帯における「習い事・塾」の機会格差
# =========================================================

# 分析に必要な変数の抽出
columns_single = [
    'w1p_配偶者の有無', # 1: いる(ふたり親), 2: いない(ひとり親)
    'w1p_習い事やスポーツクラブに通っているか', # 1: 行っている, 2: 行っていない
    'w1p_現在学習塾に行っているか' # 1: 行っている, 2: 行っていない
]

df_single = df[columns_single].copy()

# 欠損値の処理 (7777, 8888, 9999 をまとめて NaN に置換して除外)
for col in columns_single:
    df_single[col] = df_single[col].replace([7777, 8888, 9999], np.nan)

df_single = df_single.dropna()
print("有効データ数:", len(df_single))

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
# ① 世帯形態のラベル作成
df_single['family_type'] = df_single['w1p_配偶者の有無'].map({1.0: '配偶者あり（ふたり親）', 2.0: '配偶者なし（ひとり親）'})

# ② 実施フラグの作成 (1:行っている->1, 2:行っていない->0)
df_single['has_lesson'] = df_single['w1p_習い事やスポーツクラブに通っているか'].apply(lambda x: 1 if x == 1 else 0)
df_single['has_juku'] = df_single['w1p_現在学習塾に行っているか'].apply(lambda x: 1 if x == 1 else 0)

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
from IPython.display import display

# 世帯形態ごとの実施割合を算出
summary_single = df_single.groupby('family_type')[['has_lesson', 'has_juku']].mean().reset_index()

# グラフで描きやすいようにデータ形式を変換（Long形式）
df_melted_single = pd.melt(summary_single, id_vars=['family_type'], 
                           value_vars=['has_lesson', 'has_juku'],
                           var_name='activity_type', value_name='participation_rate')

# ラベルを日本語に変換
df_melted_single['activity_type'] = df_melted_single['activity_type'].map({
    'has_lesson': '習い事・スポーツクラブ', 
    'has_juku': '学習塾'
})

# 棒グラフの描画
plt.figure(figsize=(9, 6))
sns.barplot(x='activity_type', y='participation_rate', hue='family_type', data=df_melted_single, palette='Set2')
plt.title('世帯形態（ひとり親・ふたり親）による「習い事・塾」の機会格差')
plt.xlabel('活動の種類')
plt.ylabel('通っている子どもの割合')
plt.ylim(0, 1)

# グラフ上にパーセンテージを表示する工夫
ax = plt.gca()
for p in ax.patches:
    ax.annotate(f"{p.get_height():.1%}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=10, color='black', xytext=(0, 5), textcoords='offset points')

plt.legend(title='世帯形態')
plt.show()

# 数値データの表示
print("■ 分析結果：世帯形態ごとの実施割合（%）")
summary_single.columns = ['世帯形態', '習い事実施割合', '学習塾実施割合']
# 見やすくするためにパーセント表示に変換
summary_single['習い事実施割合'] = (summary_single['習い事実施割合'] * 100).round(1).astype(str) + '%'
summary_single['学習塾実施割合'] = (summary_single['学習塾実施割合'] * 100).round(1).astype(str) + '%'
display(summary_single)
```

```python
#＜ひとり親家庭では、子供の最終学歴が中学校や高校で止まってしまうのでは？＞

# =========================================================
# 仮説検証：ひとり親家庭における「進学希望（高卒以下）」の格差
# =========================================================

# 分析に必要な変数の抽出
columns_edu = [
    'w1p_配偶者の有無',        # 1: いる(ふたり親), 2: いない(ひとり親)
    'w1c_将来の希望進学段階'   # 子ども自身の進学希望
]

df_edu = df[columns_edu].copy()

# 欠損値の処理 (7777, 8888, 9999 をまとめて NaN に置換)
for col in columns_edu:
    df_edu[col] = df_edu[col].replace([7777, 8888, 9999], np.nan)

# 選択肢 7(その他) と 8(まだ決めていない) を除外して、具体的な進学目標がある層で比較
df_edu = df_edu[df_edu['w1c_将来の希望進学段階'].isin([1, 2, 3, 4, 5, 6])]
df_edu = df_edu.dropna()
print("有効データ数:", len(df_edu))

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
# ① 世帯形態のラベル作成
df_edu['family_type'] = df_edu['w1p_配偶者の有無'].map({1.0: 'ふたり親世帯', 2.0: 'ひとり親世帯'})

# ② 「中学校・高校まで（高卒以下）」を希望しているかどうかのフラグを作成
# 1: 中学校まで, 2: 高校まで -> 1 (高卒以下を希望)
# 3: 専門・各種学校, 4: 短大, 5: 大学, 6: 大学院 -> 0 (それ以上の進学を希望)
df_edu['aim_high_school_or_less'] = df_edu['w1c_将来の希望進学段階'].apply(lambda x: 1 if x in [1, 2] else 0)

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
# 世帯形態ごとの「高卒以下希望」の割合を算出
edu_summary = df_edu.groupby('family_type')['aim_high_school_or_less'].mean().reset_index()

plt.figure(figsize=(7, 5))
sns.barplot(x='family_type', y='aim_high_school_or_less', data=edu_summary, palette='Set1')
plt.title('世帯形態による「中学校・高校までの進学」を希望する子どもの割合')
plt.xlabel('世帯形態')
plt.ylabel('「中学・高校まで」を希望する割合')
plt.ylim(0, 0.5)  # 最大50%に設定（結果に応じて調整可能）

# グラフ上にパーセンテージを表示
ax = plt.gca()
for p in ax.patches:
    ax.annotate(f"{p.get_height():.1%}", (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='bottom', fontsize=12, color='black', xytext=(0, 5), textcoords='offset points')

plt.show()

# 数値データの表示
print("■ 分析結果：世帯形態ごとの「中学校・高校まで」の進学希望割合（%）")
edu_summary.columns = ['世帯形態', '中学・高校までの進学希望割合']
edu_summary['中学・高校までの進学希望割合'] = (edu_summary['中学・高校までの進学希望割合'] * 100).round(1).astype(str) + '%'
display(edu_summary)
```

```python
#＜親の人数と子供が希望する最終学歴＞

# =========================================================
# 修正版：「中高生」に絞った世帯形態と進学希望のクロス分析
# =========================================================

# 今度は「学年」の列も一緒に抽出します
columns_edu_corr = [
    'w6p_ 配偶者の有無',       # 1:いる, 2:いない
    'w6c_ 将来の希望進学段階',  
    'w6学年'                  # 1〜6:小学, 7〜9:中学, 10〜12:高校
]

df_edu_corr = df[columns_edu_corr].copy()

# 1. 「中学生・高校生（学年コード7以上）」だけにデータを絞り込む
df_edu_corr = df_edu_corr[df_edu_corr['w6学年'] >= 7]

# 2. 欠損値と未定の除外
for col in columns_edu_corr:
    df_edu_corr[col] = df_edu_corr[col].replace([7777, 8888, 9999], np.nan)
df_edu_corr['w6c_ 将来の希望進学段階'] = df_edu_corr['w6c_ 将来の希望進学段階'].replace([8, 9], np.nan)
df_edu_corr = df_edu_corr.dropna()

print(f"中高生に絞った有効データ数: {len(df_edu_corr)}人")

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
df_edu_corr['family_type'] = df_edu_corr['w6p_ 配偶者の有無'].map({1.0: 'ふたり親世帯', 2.0: 'ひとり親世帯'})

def group_education(x):
    if x in [1, 2]:
        return '1. 中学・高校まで'
    elif x in [3, 4]:
        return '2. 専門学校・短大まで'
    elif x in [5, 6, 7]:
        return '3. 大学・大学院まで'
    return np.nan

df_edu_corr['edu_aspiration'] = df_edu_corr['w6c_ 将来の希望進学段階'].apply(group_education)

# ---------------------------------------------------------
# グラフの描画
# ---------------------------------------------------------
cross_table_corr = pd.crosstab(df_edu_corr['family_type'], df_edu_corr['edu_aspiration'], normalize='index') * 100

ax = cross_table_corr.plot(kind='barh', stacked=True, figsize=(10, 4), color=['#ff9999', '#ffcc99', '#99ccff'])
plt.title('【中高生限定】世帯形態による「子どもの進学希望」の格差')
plt.xlabel('割合 (%)')
plt.ylabel('世帯形態')
plt.legend(title='子どもの希望進学段階', bbox_to_anchor=(1.05, 1), loc='upper left')

for n, x in enumerate([*cross_table_corr.index.values]):
    for (proportion, x_loc) in zip(cross_table_corr.loc[x], cross_table_corr.loc[x].cumsum()):
        if proportion > 5:
            plt.text(x=x_loc - (proportion / 2), y=n, s=f'{proportion:.1f}%', 
                     color="black", fontsize=10, ha="center", va="center")
plt.tight_layout()
plt.show()

display(round(cross_table_corr, 1))
```

```python
#＜親が忙しくて「子どもと過ごす時間」が取れないことが、子どもの「学習習慣の獲得」という機会を奪っていないか？＞

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib

# =========================================================
# 分析：親の「タイムプア」が子どもの学習習慣に与える影響
# =========================================================

# 1. 分析に必要な変数の抽出（Wave 6のデータを使用）
columns_timepoor = [
    'w6p_ あなた：ふだん子どもと過ごす時間',              # タイムプアの指標
    'w6p_ 勉強かかわり：勉強の計画の仕方を教える',        # 親のサポート
    'w6c_ 勉強方法：計画をたてて勉強する',                # 子どもの学習習慣
    'w6c_ 性格・特性：上手な勉強のしかたがわからない'     # 学習の迷子度合い
]

df_timepoor = df[columns_timepoor].copy()

# 欠損値の処理 (7777, 8888, 9999 をまとめて NaN に置換して除外)
for col in columns_timepoor:
    df_timepoor[col] = df_timepoor[col].replace([7777, 8888, 9999], np.nan)

df_timepoor = df_timepoor.dropna()
print(f"有効データ数: {len(df_timepoor)}人")

# ---------------------------------------------------------
# データの加工
# ---------------------------------------------------------
# スコアの反転（数値が大きいほど「よくする」「あてはまる」になるよう変換: 5 - 元の値）
# ※「勉強のしかたがわからない」はそのまま（数値が大きいほど「わからない」状態）
df_timepoor['teach_planning'] = 5 - df_timepoor['w6p_ 勉強かかわり：勉強の計画の仕方を教える']
df_timepoor['child_planning'] = 5 - df_timepoor['w6c_ 勉強方法：計画をたてて勉強する']
df_timepoor['dont_know_how'] = 5 - df_timepoor['w6c_ 性格・特性：上手な勉強のしかたがわからない']

# 親が子どもと過ごす時間を分かりやすくカテゴリ化（仮の分割例）
# ※実際の値の分布に合わせて調整してください（例：1〜2を短い、3を普通、4以上を長いなど）
def group_time(x):
    if x <= 2:
        return '1. 短い'
    elif x == 3:
        return '2. 普通'
    else:
        return '3. 長い'

df_timepoor['time_spent_group'] = df_timepoor['w6p_ あなた：ふだん子どもと過ごす時間'].apply(group_time)

# ---------------------------------------------------------
# グラフの描画と数値表の表示
# ---------------------------------------------------------
from IPython.display import display

# グループごとの平均値を算出
summary_timepoor = df_timepoor.groupby('time_spent_group')[['teach_planning', 'child_planning', 'dont_know_how']].mean().reset_index()

# グラフで描きやすいようにデータ形式を変換（Long形式）
df_melted_timepoor = pd.melt(summary_timepoor, id_vars=['time_spent_group'], 
                             value_vars=['teach_planning', 'child_planning', 'dont_know_how'],
                             var_name='metric_type', value_name='mean_score')

# ラベルを日本語に変換
label_map_time = {
    'teach_planning': '親：計画の立て方を教える',
    'child_planning': '子：計画を立てて勉強する',
    'dont_know_how': '子：勉強のしかたがわからない'
}
df_melted_timepoor['metric_type'] = df_melted_timepoor['metric_type'].map(label_map_time)

# 折れ線グラフの描画（傾向を見るためにLineplotを使用）
plt.figure(figsize=(10, 6))
sns.lineplot(x='time_spent_group', y='mean_score', hue='metric_type', data=df_melted_timepoor, marker='o', linewidth=2.5, markersize=10)

plt.title('親の「タイムプア（過ごす時間）」が子どもの学習習慣に与える影響')
plt.xlabel('親が子どもと過ごす時間の長さ')
plt.ylabel('平均スコア (高いほどあてはまる)')
plt.ylim(1.5, 3.5) # Y軸の範囲を見やすく調整
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='評価項目', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()

# 数値データの表示
print("■ 分析結果：親と過ごす時間の長さ別、各スコアの平均")
summary_timepoor.columns = ['親と過ごす時間の長さ', '親が計画の立て方を教える頻度', '子が計画を立てて勉強する頻度', '子が勉強のしかたがわからない度合い']
display(summary_timepoor)
```

```python
# =========================================================
# 修正版：集計案3：「進路の会話」と「行動の迷子」の関係
# =========================================================

# 列名のスペースの数（表記揺れ）を実際のデータセットに合わせました
columns_talk = [
    'w6c_  父との会話：将来や進路のこと',   # ※スペース2つ
    'w6c_  母との会話：将来や進路のこと',   # ※スペース2つ
    'w6c_ 性格・特性：上手な勉強のしかたがわからない' # ※スペース1つ
]

df_talk = df[columns_talk].copy()
df_talk = df_talk.replace([7777, 8888, 9999], np.nan)

# 父か母、どちらか会話頻度が高い方（数値が小さい方）を採用する
df_talk['talk_future'] = df_talk[['w6c_  父との会話：将来や進路のこと', 'w6c_  母との会話：将来や進路のこと']].min(axis=1)
df_talk = df_talk.dropna(subset=['talk_future', 'w6c_ 性格・特性：上手な勉強のしかたがわからない'])

# スコアの反転（数値が大きいほど「迷子になっている」状態にする）
# 元データ：1(あてはまる) 〜 4(あてはまらない)
df_talk['lost_in_learning_score'] = 5 - df_talk['w6c_ 性格・特性：上手な勉強のしかたがわからない']

talk_labels = {1.0: 'よく話す', 2.0: 'ときどき話す', 3.0: 'あまり話さない', 4.0: 'まったく話さない'}
df_talk['talk_freq'] = df_talk['talk_future'].map(talk_labels)

# グラフのX軸の並び順を指定
order_talk = ['よく話す', 'ときどき話す', 'あまり話さない', 'まったく話さない']
summary_talk = df_talk.groupby('talk_freq')['lost_in_learning_score'].mean().reindex(order_talk).reset_index()

plt.figure(figsize=(9, 5))
sns.lineplot(x='talk_freq', y='lost_in_learning_score', data=summary_talk, marker='o', linewidth=3, markersize=10, color='crimson')
plt.title('親と将来の進路について話す頻度と「勉強のしかたがわからない」度合い')
plt.xlabel('親と将来や進路について話す頻度（父母の頻度が高い方）')
plt.ylabel('学習の迷子スコア (高いほど「わからない」)')
plt.ylim(2.0, 3.5)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# 各点に数値を表示
for i in range(summary_talk.shape[0]):
    plt.text(i, summary_talk['lost_in_learning_score'][i] + 0.05, 
             f"{summary_talk['lost_in_learning_score'][i]:.2f}", 
             ha='center', va='bottom', fontsize=11, color='black')

plt.tight_layout()
plt.show()
```

```python
# =========================================================
# 仮説検証：「子どもの情報源は親である」＆「親の情報源は狭い」
# =========================================================

# ---------------------------------------------------------
# ① 子どもはどれくらい親から情報を得ているか？（進路の会話頻度）
# ---------------------------------------------------------
# Wave 7の子どもデータを使用（スペースなし）
cols_child_talk = [
    'w7c_父との会話：将来や進路のこと', 
    'w7c_母との会話：将来や進路のこと'
]
df_child = df[cols_child_talk].copy()
df_child = df_child.replace([7777, 8888, 9999], np.nan)

# 父母どちらかよく話す方の頻度を採用
df_child['talk_future_max'] = df_child[cols_child_talk].min(axis=1)

# 割合の算出
talk_counts = df_child['talk_future_max'].value_counts(normalize=True).sort_index() * 100
talk_labels = ['1. よく話す', '2. ときどき話す', '3. あまり話さない', '4. まったく話さない']

plt.figure(figsize=(6, 6))
plt.pie(talk_counts, labels=talk_labels, autopct='%1.1f%%', startangle=90, counterclock=False, colors=sns.color_palette("pastel"))
plt.title('【子ども】親と「将来や進路のこと」について話す割合 (Wave 7)')
plt.show()

# ---------------------------------------------------------
# ② 子どもの情報源である「親」は、どこから情報を得ているか？
# ---------------------------------------------------------
# Wave 7の親データを使用（スペースなし）
cols_parent_info = {
    'w7p_情報源：配偶者': '配偶者',
    'w7p_情報源：自分の親': '自分の親',
    'w7p_情報源：親戚（自分のきょうだい、おじ、おばなど）': '親戚',
    'w7p_情報源：子どもの友だちの親（ママ友・パパ友）': 'ママ友・パパ友',
    'w7p_情報源：学校の先生': '学校の先生',
    'w7p_情報源：学習塾の先生': '学習塾の先生',
    'w7p_情報源：インターネットの情報サイト': 'ネット情報サイト',
    'w7p_情報源：インターネットの掲示板・ブログ・SNS': 'ネット掲示板・SNS'
}

df_parent = df[list(cols_parent_info.keys())].copy()
df_parent = df_parent.replace([7777, 8888, 9999], np.nan)

# 各情報源を「選択した（1）」人の割合を計算
info_pct = (df_parent == 1.0).mean() * 100
info_pct.index = info_pct.index.map(cols_parent_info)
info_pct = info_pct.sort_values(ascending=False)

# グラフ化
plt.figure(figsize=(10, 6))
sns.barplot(x=info_pct.values, y=info_pct.index, palette='viridis')
plt.title('【親】子育てや教育に関する「情報源」（複数回答可・Wave 7）')
plt.xlabel('選択した親の割合 (%)')
plt.ylabel('情報源')

# 数値をバーの横に表示
for i, v in enumerate(info_pct.values):
    plt.text(v + 1, i, f"{v:.1f}%", va='center', fontsize=11)

plt.xlim(0, 100)
plt.tight_layout()
plt.show()
```

```python
# =========================================================
# 仮説検証：子どもの情報源（相談相手）は親に偏っているか？
# =========================================================

# 1. 変数の設定（Wave 7の子どもデータを使用）
cols_consult = {
    'w7c_関係：友だちに悩みを話す': '友だち',
    'w7c_関係：親に悩みを話す': '親',
    'w7c_関係：学校の先生に悩みを話す': '学校の先生'
}

# データの抽出と欠損値処理
df_consult = df[list(cols_consult.keys())].copy()
df_consult = df_consult.replace([7777, 8888, 9999], np.nan)
df_consult = df_consult.dropna()
n_sample = len(df_consult)

# 2. 「よくある」「ときどきある」と答えた割合を算出
# 元データは 1:よくある, 2:ときどきある, 3:あまりない, 4:まったくない
res_consult = {}
for col, name in cols_consult.items():
    rate = (df_consult[col] <= 2).mean() * 100
    res_consult[name] = rate

# 降順にソート
s_consult = pd.Series(res_consult).sort_values(ascending=False)

# 3. グラフ化（横棒グラフ）
plt.figure(figsize=(9, 5))
sns.barplot(x=s_consult.values, y=s_consult.index, palette='Set2')

plt.title('【子ども】悩みを話す相手の割合（「よくある」「ときどきある」の合計） (Wave 7)', fontsize=14)
plt.xlabel('相談する割合 (%)', fontsize=12)
plt.xlim(0, 100)

# 各バーに数値を表示
for i, v in enumerate(s_consult.values):
    plt.text(v + 1, i, f"{v:.1f}%", va='center', fontsize=12, color='black', fontweight='bold')

# 右下にN数を表示
plt.figtext(0.99, 0.01, f"N={n_sample}", ha='right', fontsize=12, color='gray')

plt.tight_layout()
plt.show()
```

```python
# =========================================================
# 仮説検証：親の情報源の豊富さと、子どもの目標・意欲の関係
# =========================================================

# 1. 変数の設定（Wave 7を使用）
# 親の各種情報源
cols_info = [
    'w7p_情報源：配偶者', 'w7p_情報源：自分の子ども', 'w7p_情報源：自分の親',
    'w7p_情報源：配偶者の親', 'w7p_情報源：親戚（自分のきょうだい、おじ、おばなど）',
    'w7p_情報源：子どもの友だちの親（ママ友・パパ友）', 'w7p_情報源：友人',
    'w7p_情報源：職場の知人', 'w7p_情報源：学校の先生', 'w7p_情報源：学習塾の先生',
    'w7p_情報源：テレビ', 'w7p_情報源：新聞', 'w7p_情報源：書籍（専門書など）',
    'w7p_情報源：雑誌（専門誌など）', 'w7p_情報源：インターネットの情報サイト',
    'w7p_情報源：インターネットの掲示板・ブログ・SNS'
]
# 子どもの意識
col_goal = 'w7c_性格・特性：将来の目標がはっきりしている'
col_challenge = 'w7c_性格・特性：難しいことや新しいことにいつも挑戦したい'

# データ抽出と欠損値処理
df_analysis = df[cols_info + [col_goal, col_challenge]].copy()
df_analysis = df_analysis.replace([7777, 8888, 9999], np.nan)

# 2. 親の情報源の数をカウント（1.0＝選択した数を横方向に合計）
df_analysis['info_count'] = (df_analysis[cols_info] == 1.0).sum(axis=1)

# 3. 子どものスコアを反転（数値が大きいほど「あてはまる（意欲的）」にする）
df_analysis['goal_score'] = 5 - df_analysis[col_goal]
df_analysis['challenge_score'] = 5 - df_analysis[col_challenge]

# 子どものスコアが有効なデータのみを残し、N数を確定
df_analysis = df_analysis.dropna(subset=['goal_score', 'challenge_score'])
n_sample = len(df_analysis)

# 情報源の数をグルーピング（見やすく分類）
def group_info_count(x):
    if x <= 1: return '0-1個'
    elif x <= 3: return '2-3個'
    elif x <= 5: return '4-5個'
    else: return '6個以上'

df_analysis['info_group'] = df_analysis['info_count'].apply(group_info_count)
order_info = ['0-1個', '2-3個', '4-5個', '6個以上']

# グループごとの平均スコアを算出
summary = df_analysis.groupby('info_group')[['goal_score', 'challenge_score']].mean().reindex(order_info).reset_index()

# 4. グラフ化（2軸の折れ線グラフ）
plt.figure(figsize=(9, 6))

plt.plot(summary['info_group'], summary['goal_score'], marker='o', linewidth=3, markersize=10, label='将来の目標がはっきりしている', color='royalblue')
plt.plot(summary['info_group'], summary['challenge_score'], marker='s', linewidth=3, markersize=10, label='新しいことに挑戦したい', color='darkorange')

plt.title('親の「情報源の数」と子どもの「目標・挑戦意欲」の関係 (Wave 7)')
plt.xlabel('親が活用している情報源の数')
plt.ylabel('子どもの意識スコア平均 (高いほどあてはまる)')
plt.ylim(2.0, 3.5)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='子どもの性格・特性')

# 各点に数値を表示
for i in range(summary.shape[0]):
    plt.text(i, summary['goal_score'][i] + 0.04, f"{summary['goal_score'][i]:.2f}", ha='center', va='bottom', fontsize=11, color='royalblue')
    plt.text(i, summary['challenge_score'][i] - 0.08, f"{summary['challenge_score'][i]:.2f}", ha='center', va='top', fontsize=11, color='darkorange')

# 右下にN数をちょこんと表示
plt.figtext(0.99, 0.01, f"N={n_sample}", ha='right', fontsize=12, color='gray')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import japanize_matplotlib

# =========================================================
# 仮説検証：「身近な働く大人からの情報」と「進路の具体性」の関係
# =========================================================

# 1. 変数の設定（Wave 7を使用）
col_work_story = 'w7c_経験：親から仕事の楽しさや大変さを聞く'
col_goal_clear = 'w7c_性格・特性：将来の目標がはっきりしている'
col_has_dream = 'w7c_将来なりたい職業はあるか'

cols_hypothesis = [col_work_story, col_goal_clear, col_has_dream]

# データの抽出と欠損値処理
df_analysis = df[cols_hypothesis].copy()
df_analysis = df_analysis.replace([7777, 8888, 9999], np.nan)
df_analysis = df_analysis.dropna()
n_sample = len(df_analysis)

# 2. データの加工
# 経験の有無をわかりやすいラベルに変換（1.0: 選択＝聞いたことがある, 0.0: 非選択＝聞いたことがない）
df_analysis['work_story_exp'] = df_analysis[col_work_story].map({1.0: 'ある', 0.0: 'ない'})

# 目標の明確さスコアを反転（数値が大きいほど「あてはまる（目標が明確）」にする）
# 元データ: 1(とてもあてはまる) 〜 4(まったくあてはまらない)
df_analysis['goal_score'] = 5 - df_analysis[col_goal_clear]

# なりたい職業が「ある(1.0)」人の割合を計算するためのダミー変数
df_analysis['has_dream_flag'] = df_analysis[col_has_dream].apply(lambda x: 1 if x == 1.0 else 0)

# 3. グループごとの集計
summary_goal = df_analysis.groupby('work_story_exp')['goal_score'].mean().reset_index()
summary_dream = df_analysis.groupby('work_story_exp')['has_dream_flag'].mean().reset_index()
summary_dream['has_dream_pct'] = summary_dream['has_dream_flag'] * 100

# 4. グラフ化（2つのグラフを並べて表示）
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# グラフ1：仕事の話を聞く経験と「将来の目標」の明確さ
sns.barplot(x='work_story_exp', y='goal_score', data=summary_goal, ax=ax1, palette='Set2', order=['ある', 'ない'])
ax1.set_title('「親から仕事の話を聞く経験」と将来の目標の明確さ', fontsize=12)
ax1.set_xlabel('親から仕事の楽しさや大変さを聞く経験')
ax1.set_ylabel('将来の目標がはっきりしている (平均スコア)')
ax1.set_ylim(2.0, 3.5)

for p in ax1.patches:
    ax1.annotate(f"{p.get_height():.2f}", (p.get_x() + p.get_width() / 2., p.get_height()), 
                 ha='center', va='bottom', fontsize=11, xytext=(0, 5), textcoords='offset points')

# グラフ2：仕事の話を聞く経験と「なりたい職業」の有無
sns.barplot(x='work_story_exp', y='has_dream_pct', data=summary_dream, ax=ax2, palette='Set2', order=['ある', 'ない'])
ax2.set_title('「親から仕事の話を聞く経験」となりたい職業の有無', fontsize=12)
ax2.set_xlabel('親から仕事の楽しさや大変さを聞く経験')
ax2.set_ylabel('将来なりたい職業がある割合 (%)')
ax2.set_ylim(0, 100)

for p in ax2.patches:
    ax2.annotate(f"{p.get_height():.1f}%", (p.get_x() + p.get_width() / 2., p.get_height()), 
                 ha='center', va='bottom', fontsize=11, xytext=(0, 5), textcoords='offset points')

# N数表示
plt.figtext(0.98, 0.02, f"N={n_sample}", ha='right', fontsize=11, color='gray')

plt.tight_layout()
plt.show()
```
