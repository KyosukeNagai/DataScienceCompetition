```python
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib
import os

japanize_matplotlib.japanize()
file_path = "../Database/Benesse/1571.csv"

df = pd.read_csv(file_path)
display(df.head())
```

```python
len(df.index)
```

```python
label_path = "../Database/Benesse/1571.label.txt"
label = pd.read_table(label_path, encoding="utf-8-sig", header=0)
display(label)

# 4028行目までのデータを使用する
label = label.iloc[2:4028+1]

# Unnamed2 を日本語カラム名に変更する
label = label.rename(columns={
    "Unnamed: 2": "ラベル",
    "Unnamed: 1": "変数"
})
display(label)
```

```python
rename_map = {
    row["変数情報"]: row["ラベル"]
    for _, row in label.iterrows()
    if pd.notna(row.get("変数情報")) and pd.notna(row.get("ラベル"))
}
```

```python
df_ja = df.rename(columns=rename_map)

display(df_ja)
```

```python
df_ja.to_csv("../Database/Benesse/1571_ja.csv", index=False, encoding="utf-8-sig")
```

```python
df_ja.to_parquet("../Database/Benesse/1571_ja.parquet", index=False, engine="pyarrow")
```
