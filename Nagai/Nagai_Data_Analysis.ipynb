```python
import re
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns


# 学年段階の定義
def stage_name(grade):
    if grade in [1, 2, 3, 4, 5, 6]:
        return "小学生"
    elif grade in [7, 8, 9]:
        return "中学生"
    elif grade in [10, 11, 12]:
        return "高校生"


# 並び順を明示的に定義
stage_order = ["小学生", "中学生", "高校生"]

# research_flag に入っている w1 系列を、全 wave へ展開してまとめて描画
base_keys = []
for col in df_w1.columns:
    m = re.match(r"^w\d+([cp])_(.+)$", col)
    if m:
        base_keys.append((m.group(1), m.group(2)))

# 重複を除く
seen = set()
base_keys = [
    (prefix, suffix)
    for prefix, suffix in base_keys
    if not ((prefix, suffix) in seen or seen.add((prefix, suffix)))
]

use_df = df

for prefix, suffix in base_keys:
    cols = [
        c
        for c in use_df.columns
        if re.match(rf"^w\d+{prefix}_{re.escape(suffix)}$", c)
    ]
    if not cols:
        continue

    frames = []
    for col in cols:
        m = re.match(r"^w(\d+)([cp])_(.+)$", col)
        if not m:
            continue
        wave = int(m.group(1))
        grade_col = f"w{wave}学年"
        if grade_col not in use_df.columns:
            continue

        frames.append(
            pd.DataFrame({
                "wave": wave,
                "stage": use_df[grade_col].map(stage_name),
                "value": use_df[col],
            })
        )

    if not frames:
        continue

    tmp = pd.concat(frames, ignore_index=True)
    tmp = tmp.dropna(subset=["value", "stage"])
    tmp = tmp[~tmp["value"].isin([7777, 8888, 9999])]
    if tmp.empty:
        continue

    # col_order パラメータで並び順を指定
    if (
        pd.api.types.is_numeric_dtype(tmp["value"])
        and tmp["value"].nunique() > 10
    ):
        g = sns.displot(
            data=tmp,
            x="value",
            col="stage",
            col_order=stage_order,  # ← ここを追加
            hue="wave",
            kind="hist",
            bins=10,
            kde=True,
            height=3.5,
            aspect=1.2,
        )
    else:
        g = sns.catplot(
            data=tmp,
            x="value",
            col="stage",
            col_order=stage_order,  # ← ここを追加
            hue="wave",
            kind="count",
            height=3.5,
            aspect=1.2,
        )

    g.fig.suptitle(f"{prefix}_{suffix}", y=1.02)
    plt.show()
```

```python
import re
import numpy as np
import pandas as pd
import scipy.stats as stats
from IPython.display import display

# クラメールのV係数を計算する関数
def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = stats.chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    
    # ゼロ除算エラーの回避
    r, k = confusion_matrix.shape
    if n == 0 or min(k-1, r-1) == 0:
        return np.nan
        
    phi2 = chi2 / n
    return np.sqrt(phi2 / min(k-1, r-1))

# ターゲット変数のベース部分（ウェーブ番号 'w1' を除いた部分）を指定
target_base = "c_経験：地域の行事に参加する（夏祭りなど）"

# データフレームから存在するウェーブ番号を取得（w1, w2, w3...）
waves = set()
for col in df.columns:
    m = re.match(r"^w(\d+)", col)
    if m:
        waves.add(int(m.group(1)))
waves = sorted(list(waves))

results = []

# ウェーブごとにループ処理
for wave in waves:
    target_col = f"w{wave}{target_base}"

    # そのウェーブにターゲット変数が存在しない場合はスキップ（例: 特定の年だけ質問されなかった場合）
    if target_col not in df.columns:
        continue

    # 同じウェーブの変数のみを比較対象とする（例：w1のターゲットにはw1の変数のみ比較）
    wave_cols = [c for c in df.columns if re.match(rf"^w{wave}", c)]

    for col in wave_cols:
        if col == target_col:
            continue

        # 2変数のデータフレームを作成し、欠損値(NaN)を除外
        tmp = df[[target_col, col]].dropna()

        # 特殊な回答コード（わからない、非該当、無回答など）を除外
        # ※データ定義書に合わせて除外数値を調整してください
        exclude_vals = [7777, 8888, 9999]
        tmp = tmp[~tmp[target_col].isin(exclude_vals)]
        tmp = tmp[~tmp[col].isin(exclude_vals)]

        # データが少なすぎる、または値が1種類しかない場合はスキップ
        if len(tmp) < 10 or tmp[col].nunique() < 2:
            continue

        # 1. スピアマンの順位相関係数
        try:
            spearman_corr, _ = stats.spearmanr(tmp[target_col], tmp[col])
        except:
            spearman_corr = np.nan

        # 2. クラメールのV係数
        cramer_v_val = cramers_v(tmp[target_col], tmp[col])

        results.append({
            "wave": wave,
            "比較変数": col,
            "スピアマン順位相関": spearman_corr,
            "絶対値_スピアマン": abs(spearman_corr) if not pd.isna(spearman_corr) else 0,
            "クラメールV係数": cramer_v_val
        })

# 結果をデータフレーム化
result_df = pd.DataFrame(results)

# ウェーブごとに結果を出力
for wave in waves:
    wave_df = result_df[result_df["wave"] == wave]
    if wave_df.empty:
        continue

    print(f"\n{'='*60}")
    print(f"Wave {wave} の分析結果 (ターゲット: w{wave}{target_base})")
    print(f"{'='*60}")

    print("\n■ スピアマンの順位相関（順序の連動性が強い変数）")
    display(wave_df.sort_values("絶対値_スピアマン", ascending=False)[["比較変数", "スピアマン順位相関"]])

    print("\n■ クラメールのV係数（カテゴリの偏りが強い変数）")
    display(wave_df.sort_values("クラメールV係数", ascending=False)[["比較変数", "クラメールV係数"]])
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm

# --------------------------------------------------
# 1. 変数の設定とデータの前処理
# --------------------------------------------------
# ※実際のデータフレームのカラム名に合わせて書き換えてください
target_col = "w1c_経験：地域の行事に参加する（夏祭りなど）"
conv_col = "w1c_父との会話：社会のニュース"
smartphone_col = "w1c_生活時間：テレビゲームや携帯ゲーム機で遊ぶ"

# 分析用のデータフレームを作成し、無効回答（7777, 8888, 9999など）を欠損値(NaN)に変換
df_model = df[[target_col, conv_col, smartphone_col]].copy()
for col in df_model.columns:
    df_model[col] = df_model[col].replace([7777, 8888, 9999], np.nan)

# 欠損値を含む行を削除（完全なデータのみでモデルを作るため）
df_model = df_model.dropna()


# --------------------------------------------------
# 2. 目的変数（Y）の2値化（0 or 1）
# --------------------------------------------------
# 例：1(よくする), 2(たまにする) を「1:調べる層」
#   3(あまりしない), 4(まったくしない) を「0:調べない層」とする
df_model["Y_調べる行動"] = df_model[target_col].apply(lambda x: 1 if x <= 2 else 0)


# --------------------------------------------------
# 3. 説明変数（X）の方向調整
# --------------------------------------------------
# アンケートは「1が最もポジティブ（よくする）」であることが多いため、
# 「数値が大きいほど行動頻度が高い／時間が長い」となるように反転・調整します。

X = pd.DataFrame()

# 親との会話（1:よくする ～ 4:まったくしない の場合、5から引いて逆転させる）
# -> これにより、4:よくする, 3:たまにする, 2:あまりしない, 1:まったくしない になる
X["親とニュースの話をする"] = 5 - df_model[conv_col]

# スマホ利用時間（選択肢の数字が大きいほど時間が長い前提。そのまま使用）
X["スマホ利用時間"] = df_model[smartphone_col]

# モデルの切片（定数項）を追加
X = sm.add_constant(X)


# --------------------------------------------------
# 4. ロジスティック回帰モデルの実行
# --------------------------------------------------
# YとXを指定して学習
model = sm.Logit(df_model["Y_調べる行動"], X)
result = model.fit(method='bfgs')

# 結果のサマリー（詳細な統計情報）を表示したい場合は以下を実行
# print(result.summary())


# --------------------------------------------------
# 5. オッズ比とp値の抽出
# --------------------------------------------------
# 係数（パラメータ）を指数関数（np.exp）にかけて「オッズ比」に変換
odds_ratios = np.exp(result.params)

# 95%信頼区間もオッズ比に変換
conf = np.exp(result.conf_int())
conf.columns = ['95%信頼区間_下限', '95%信頼区間_上限']

# 結果を見やすいデータフレームにまとめる
summary_df = pd.DataFrame({
    'オッズ比': odds_ratios,
    'p値': result.pvalues,
    '95%信頼区間_下限': conf['95%信頼区間_下限'],
    '95%信頼区間_上限': conf['95%信頼区間_上限']
})

# 切片（const）は解釈に不要なため除外して表示
print("■ 情報探索行動を生み出す要因（ロジスティック回帰分析）")
display(summary_df.drop('const'))
```

```python
import re
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency
from IPython.display import display

# ターゲット変数のベース部分を指定
target_base = "c_性格・特性：難しいことや新しいことにいつも挑戦したい"

# 除外する特殊な回答コード
missing_codes = [7777, 8888, 9999]

# クロス集計とカイ二乗検定を行う関数
def chi_square_summary(data, target, feature):
    pair = data[[target, feature]].copy()
    pair = pair.dropna()
    pair = pair[~pair[target].isin(missing_codes)]
    pair = pair[~pair[feature].isin(missing_codes)]

    if pair.empty:
        return None

    ct = pd.crosstab(pair[target], pair[feature])
    # 2x2以上のクロス表になっていない場合は検定不可のため除外
    if ct.shape[0] < 2 or ct.shape[1] < 2:
        return None

    chi2, p_value, dof, expected = chi2_contingency(ct)
    n = ct.to_numpy().sum()
    r, k = ct.shape
    cramers_v = np.sqrt((chi2 / n) / min(r - 1, k - 1)) if min(r - 1, k - 1) > 0 else np.nan

    return {
        "変数": feature,
        "サンプル数": int(n),
        "自由度": int(dof),
        "chi2": chi2,
        "p値": p_value,
        "Cramér's V": cramers_v,
        "クロス表": ct,
    }

# データフレーム (df_answer) から存在するウェーブ番号を取得（w1, w2, w3...）
waves = set()
for col in df.columns:
    m = re.match(r"^w(\d+)", col)
    if m:
        waves.add(int(m.group(1)))
waves = sorted(list(waves))

print(f"分析対象のウェーブ: {waves}")

# ウェーブごとにループ処理
for wave in waves:
    target_col = f"w{wave}{target_base}"

    # そのウェーブにターゲット変数が存在しない場合はスキップ
    if target_col not in df.columns:
        print(f"【スキップ】Wave {wave} にはターゲット変数 ({target_col}) が存在しません。") # ← この1行を追加
        continue

    # 検索したい情報のキーワードを指定（実際の調査票に合わせて調整してください）
    info_keywords = ["情報", "SNS", "ネット", "回覧板", "広報", "知る", "関心", "積極", "スマートフォン", "スマホ", "携帯", "テレビ", "新聞", "雑誌", "書籍", "本", "パソコン", "地域", "将来", "行動", "体験", "挑戦", "進路"]

    candidate_cols = []
    for col in df.columns:
        # 条件を and で1つの if 文に結合（カッコを使って改行すると読みやすくなります）
        if (re.match(rf"^w{wave}", col) 
            and col != target_col 
            and not col.startswith(f"w{wave}回答フラグ") 
            and any(kw in col for kw in info_keywords)):
            candidate_cols.append(col)

    results = []
    # 候補変数ごとにカイ二乗検定を実行
    for col in candidate_cols:
        summary = chi_square_summary(df, target_col, col)
        if summary is not None:
            results.append(summary)

    # 出力のフォーマット
    print(f"\n{'='*80}")
    print(f"Wave {wave} のカイ二乗検定・クロス集計結果 (ターゲット: {target_col})")
    print(f"{'='*80}")

    chi_df = pd.DataFrame(results)

    if chi_df.empty:
        print("有効なクロス集計対象がありませんでした。")
    else:
        # Cramér's V の降順、p値の昇順で並べ替え
        sorted_chi_df = chi_df.sort_values(["Cramér's V", "p値"], ascending=[False, True])

        sorted_chi_df = sorted_chi_df[sorted_chi_df["p値"] < 0.10]

        # データフレームを見やすくフォーマットして表示
        display(
            sorted_chi_df[["変数", "サンプル数", "自由度", "chi2", "p値", "Cramér's V"]]
            .style.format({"chi2": "{:.3f}", "p値": "{:.3e}", "Cramér's V": "{:.3f}"})
        )
```

```python
# ==============================================================================
# 0. 分析設定フラグ
# ==============================================================================
# True : 重回帰分析（一括）のみを実行
# False: 単回帰分析（1変数）と重回帰分析（一括）の両方を実行
ONLY_MULTIPLE_REGRESSION = True


# ==============================================================================
# 2. 変数テンプレートの設定
# ==============================================================================
target_var = "w{wave}c_性格・特性：難しいことや新しいことにいつも挑戦したい"

binary_vars = [
    "w{wave}c_経験：地域の行事に参加する（夏祭りなど）",
    "w{wave}c_将来なりたい職業はあるか",
]

continuous_vars = [
    "w{wave}c_生活時間：本を読む",
    "w{wave}p_得意・苦手：自分で決めて行動すること",
    "w{wave}c_得意・苦手：いろいろな情報から信頼できるものを選んで使うこと",
    "w{wave}c_学校生活：授業の話し合いで積極的に発言する",
    "w{wave}c_学習意欲：新しいことを知るのがうれしいから",
    "w{wave}c_性格・特性：将来の目標がはっきりしている",
    "w{wave}c_学習意欲：将来なりたい職業につきたいから",
    "w{wave}p_教育意識：子どもには今のうちにいろいろな体験をさせたい",
    "w{wave}p_大切さを教える：将来の目標をもつこと",
]


# ==============================================================================
# 3. 学校段階 (小学生・中学生・高校生) × Wave (1〜7) でのループ分析
# ==============================================================================
stage_order = ["小学生", "中学生", "高校生"]
all_results = []

for stage in stage_order:
    for wave in range(1, 8):
        grade_col = f"w{wave}学年"
        y_col = target_var.format(wave=wave)

        # 学年カラム、目的変数がデータに含まれていない場合はスキップ
        if (
            grade_col not in df_event.columns
            or y_col not in df_event.columns
        ):
            continue

        # --- 学校段階によるデータのフィルタリング ---
        if stage == "小学生":
            stage_df = df_event[df_event[grade_col].isin(range(1, 7))]
        elif stage == "中学生":
            stage_df = df_event[df_event[grade_col].isin(range(7, 10))]
        elif stage == "高校生":
            stage_df = df_event[df_event[grade_col].isin(range(10, 13))]

        if stage_df.empty:
            continue

        # --- 目的変数 (y) のクリーニングと検証 ---
        y_raw = pd.to_numeric(stage_df[y_col], errors="coerce")
        u_vals = sorted(y_raw.dropna().unique())

        if len(u_vals) != 2:
            continue

        y_series = y_raw.map({u_vals[0]: 0, u_vals[1]: 1})

        # --- 存在する説明変数の抽出し標準化 ---
        b_cols = [
            f.format(wave=wave)
            for f in binary_vars
            if f.format(wave=wave) in stage_df.columns
        ]
        c_cols = [
            f.format(wave=wave)
            for f in continuous_vars
            if f.format(wave=wave) in stage_df.columns
        ]

        if not b_cols and not c_cols:
            continue

        X_dict = {}
        for col in b_cols:
            X_dict[col] = stage_df[col]

        for col in c_cols:
            std_val = stage_df[col].std()
            if std_val != 0 and pd.notnull(std_val):
                X_dict[f"{col}_std"] = (
                    stage_df[col] - stage_df[col].mean()
                ) / std_val
            else:
                X_dict[col] = stage_df[col]

        X_all = pd.DataFrame(X_dict)
        feature_names = list(X_all.columns)

        # --------------------------------------------------
        # A. 単回帰分析（1変数ずつ）※ フラグが False の場合のみ実行
        # --------------------------------------------------
        if not ONLY_MULTIPLE_REGRESSION:
            for col in feature_names:
                X_single = sm.add_constant(X_all[[col]])
                valid_idx = y_series.dropna().index.intersection(
                    X_single.dropna().index
                )

                if len(valid_idx) < 10:
                    continue

                try:
                    model_s = sm.Logit(
                        y_series.loc[valid_idx], X_single.loc[valid_idx]
                    ).fit(disp=False)

                    res_s = pd.DataFrame(
                        {
                            "School Stage": stage,
                            "Wave": f"Wave {wave}",
                            "Model": "単回帰 (1変数)",
                            "Variable": model_s.params.index,
                            "Odds Ratio": np.exp(model_s.params),
                            "p-value": model_s.pvalues,
                            "2.5%": np.exp(model_s.conf_int()[0]),
                            "97.5%": np.exp(model_s.conf_int()[1]),
                        }
                    ).reset_index(drop=True)

                    all_results.append(res_s)
                except Exception as e:
                    pass

        # --------------------------------------------------
        # B. 重回帰分析（一括）
        # --------------------------------------------------
        if len(feature_names) >= 2:
            X_multi = sm.add_constant(X_all)
            valid_idx = y_series.dropna().index.intersection(
                X_multi.dropna().index
            )

            if len(valid_idx) >= 10:
                try:
                    model_m = sm.Logit(
                        y_series.loc[valid_idx], X_multi.loc[valid_idx]
                    ).fit(disp=False)

                    res_m = pd.DataFrame(
                        {
                            "School Stage": stage,
                            "Wave": f"Wave {wave}",
                            "Model": "重回帰 (一括)",
                            "Variable": model_m.params.index,
                            "Odds Ratio": np.exp(model_m.params),
                            "p-value": model_m.pvalues,
                            "2.5%": np.exp(model_m.conf_int()[0]),
                            "97.5%": np.exp(model_m.conf_int()[1]),
                        }
                    ).reset_index(drop=True)

                    all_results.append(res_m)
                except Exception as e:
                    pass


# ==============================================================================
# 4. 全結果の結合・カラム名短縮・スタイル装飾
# ==============================================================================
if all_results:
    final_results = pd.concat(all_results, ignore_index=True)

    # 2. 有意水準記号 (*) を付与したオッズ比列を作成
    def format_or_with_stars(row):
        p = row["p-value"]
        or_val = row["Odds Ratio"]
        if p < 0.001:
            star = "***"
        elif p < 0.01:
            star = "**"
        elif p < 0.05:
            star = "*"
        else:
            star = ""
        return f"{or_val:.3f}{star}"

    final_results["Odds Ratio (Sig)"] = final_results.apply(
        format_or_with_stars, axis=1
    )

    # 表示列の並び順を設定し、カラム名をコンパクトに変更
    display_cols = [
        "School Stage",
        "Wave",
        "Model",
        "Variable",
        "Odds Ratio (Sig)",
        "p-value",
        "2.5%",
        "97.5%",
    ]
    df_show = final_results[display_cols].rename(
        columns={
            "Odds Ratio (Sig)": "OR",
            "97.5%": "97.5%",
        }
    )

# 3. スタイリング関数の定義
    def apply_custom_styles(styler):
        styler.set_table_styles(
            [
                {
                    "selector": "th",
                    "props": [
                        ("background-color", "#f2f4f7"),
                        ("color", "#333333"),
                        ("font-weight", "bold"),
                        ("text-align", "center"),
                    ],
                },
                {
                    "selector": "td",
                    "props": [
                        ("white-space", "nowrap"),
                        ("padding", "6px 12px"),
                    ],
                },
                # Wave列（1列目）の文字を太字・中央揃えにして色付け
                {
                    "selector": "th.col1, td.col1",
                    "props": [
                        ("text-align", "center"),
                        ("font-weight", "bold"),
                    ],
                },
                # OR列（4列目）と 97.5%列（7列目）の幅を固定
                {
                    "selector": "th.col4, td.col4",
                    "props": [
                        ("width", "70px"),
                        ("max-width", "70px"),
                        ("text-align", "right"),
                    ],
                },
                {
                    "selector": "th.col7, td.col7",
                    "props": [
                        ("width", "60px"),
                        ("max-width", "60px"),
                        ("text-align", "right"),
                    ],
                },
            ]
        )

        # --------------------------------------------------
        # 特定の列（Wave・const・有意行）へのスタイル適用関数
        # --------------------------------------------------
        # Wave番号に応じた背景パステルカラーのマップ設定
        wave_colors = {
            "Wave 1": "#e3f2fd",  # 淡い青
            "Wave 2": "#e8f5e9",  # 淡い緑
            "Wave 3": "#fff3e0",  # 淡いオレンジ
            "Wave 4": "#f3e5f5",  # 淡い紫
            "Wave 5": "#fffde7",  # 淡い黄
            "Wave 6": "#e0f2f1",  # 淡いティール
            "Wave 7": "#fbe9e7",  # 淡いサーモン
        }

        def style_rows(row):
            is_p_sig = (
                row["p-value"] < 0.05 if pd.notnull(row["p-value"]) else False
            )
            is_const = row["Variable"] == "const"
            wave_val = row["Wave"]

            # 行全体のスタイル生成
            if is_const:
                styles = ["color: #888888; font-style: italic;"] * len(row)
            elif is_p_sig:
                styles = [
                    "background-color: #e6f4ea; font-weight: bold;"
                ] * len(row)
            else:
                styles = [""] * len(row)

            # Wave列（列インデックス: 1）のみ個別に文字色・背景色を指定
            wave_bg = wave_colors.get(wave_val, "#ffffff")
            # Wave列の文字色（濃い青で強調）＋背景色
            styles[1] = (
                f"background-color: {wave_bg}; color: #1565c0; font-weight: bold;"
            )

            return styles

        styler = styler.apply(style_rows, axis=1)

        # インデックス行を非表示
        styler = styler.hide(axis="index")

        styler = styler.format(
            {"p-value": "{:.4f}", "2.5%": "{:.3f}", "97.5%": "{:.3f}"}
        )

        return styler

        def highlight_significant(row):
            is_p_sig = (
                row["p-value"] < 0.05 if pd.notnull(row["p-value"]) else False
            )
            is_const = row["Variable"] == "const"

            if is_const:
                return ["color: #888888; font-style: italic;"] * len(row)
            elif is_p_sig:
                return [
                    "background-color: #e6f4ea; font-weight: bold;"
                ] * len(row)
            else:
                return [""] * len(row)

        styler = styler.apply(highlight_significant, axis=1)

        # インデックス行を非表示
        styler = styler.hide(axis="index")

        styler = styler.format(
            {"p-value": "{:.4f}", "2.5%": "{:.3f}", "97.5%": "{:.3f}"}
        )

        return styler

    # スタイル適用
    styled_results = apply_custom_styles(df_show.style)
    display(styled_results)

else:
    print("条件に合致するデータがなく、分析を実行できませんでした。")
```
